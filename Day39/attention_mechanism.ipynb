```python
import torch
import torch.nn.functional as F
```

```python
Q = torch.tensor([[1.0, 0.0]])

K = torch.tensor([
    [1.0, 0.0],
    [0.0, 1.0],
    [1.0, 1.0]
])

V = torch.tensor([
    [10.0, 0.0],
    [0.0, 10.0],
    [5.0, 5.0]
])
```

```python
scores = Q @ K.T

print(scores)
```

```python
attention_weights = F.softmax(scores, dim=-1)

print(attention_weights)
```

```python
output = attention_weights @ V

print(output)
```

```python
d_k = K.shape[-1]

scores = (Q @ K.T) / torch.sqrt(
    torch.tensor(d_k, dtype=torch.float32)
)
```

```python
attention_weights = F.softmax(scores, dim=-1)

output = attention_weights @ V

print(output)
```
