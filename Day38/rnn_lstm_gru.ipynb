```python
import torch
import torch.nn as nn
```

```python
rnn = nn.RNN(
    input_size=10,
    hidden_size=20,
    batch_first=True
)

x = torch.randn(2, 5, 10)

output, hidden = rnn(x)

print("Output shape:", output.shape)
print("Hidden shape:", hidden.shape)
```

```python
lstm = nn.LSTM(
    input_size=10,
    hidden_size=20,
    batch_first=True
)

output, (hidden, cell) = lstm(x)

print("Output shape:", output.shape)
print("Hidden shape:", hidden.shape)
print("Cell shape:", cell.shape)
```

```python
gru = nn.GRU(
    input_size=10,
    hidden_size=20,
    batch_first=True
)

output, hidden = gru(x)

print("Output shape:", output.shape)
print("Hidden shape:", hidden.shape)
```

```python
rnn = nn.RNN(10, 20, batch_first=True)
lstm = nn.LSTM(10, 20, batch_first=True)
gru = nn.GRU(10, 20, batch_first=True)

rnn_output, rnn_hidden = rnn(x)

lstm_output, (lstm_hidden, lstm_cell) = lstm(x)

gru_output, gru_hidden = gru(x)

print("RNN:", rnn_output.shape)
print("LSTM:", lstm_output.shape)
print("GRU:", gru_output.shape)
```
