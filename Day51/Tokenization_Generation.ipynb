```python
from transformers import AutoTokenizer
```

```python
tokenizer = AutoTokenizer.from_pretrained("gpt2")
```

```python
texts = [
    "playing",
    "unhappiness",
    "machine learning",
    "internationalization"
]

for text in texts:
    tokens = tokenizer.tokenize(text)
    
    print("Text:", text)
    print("Tokens:", tokens)
    print("-" * 40)
```

```python
text = "Machine learning is interesting."

tokens = tokenizer.tokenize(text)
token_ids = tokenizer.encode(text)

print("Tokens:")
print(tokens)

print("\nToken IDs:")
print(token_ids)
```

```python
from transformers import pipeline
```

```python
generator = pipeline(
    "text-generation",
    model="distilgpt2"
)
```

```python
prompt = "Artificial intelligence will"

result = generator(
    prompt,
    max_new_tokens=40,
    temperature=0.2,
    do_sample=True
)

print(result[0]["generated_text"])
```

```python
result = generator(
    prompt,
    max_new_tokens=40,
    temperature=0.7,
    do_sample=True
)

print(result[0]["generated_text"])
```

```python
result = generator(
    prompt,
    max_new_tokens=40,
    temperature=1.2,
    do_sample=True
)

print(result[0]["generated_text"])
```

```python
result = generator(
    prompt,
    max_new_tokens=40,
    do_sample=True,
    top_k=5,
    temperature=0.7
)

print(result[0]["generated_text"])
```

```python
settings = [
    {"temperature": 0.2, "top_k": 20, "top_p": 0.9},
    {"temperature": 0.7, "top_k": 20, "top_p": 0.9},
    {"temperature": 1.2, "top_k": 20, "top_p": 0.9}
]

for setting in settings:

    result = generator(
        prompt,
        max_new_tokens=30,
        do_sample=True,
        temperature=setting["temperature"],
        top_k=setting["top_k"],
        top_p=setting["top_p"]
    )

    print("Temperature:", setting["temperature"])
    print(result[0]["generated_text"])
    print("-" * 70)
```
