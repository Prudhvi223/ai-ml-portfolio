```python
from transformers import pipeline
```

```python
sentiment1 = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)
```

```python
text = "I really enjoyed this movie. It was amazing!"

result1 = sentiment1(text)

print(result1)
```

```python
texts = [
    "I love this product!",
    "This is the worst experience ever.",
    "The food was okay.",
    "I am extremely happy today."
]

for text in texts:
    result = sentiment1(text)
    print(text)
    print(result)
    print()
```

```python
sentiment2 = pipeline(
    "sentiment-analysis",
    model="cardiffnlp/twitter-roberta-base-sentiment-latest"
)
```

```python
texts = [
    "I love this product!",
    "This is the worst experience ever.",
    "The food was okay.",
    "I am extremely happy today."
]

for text in texts:

    result1 = sentiment1(text)
    result2 = sentiment2(text)

    print("Text:", text)
    print("Model 1:", result1)
    print("Model 2:", result2)
    print("-" * 50)
```

```python
for text in texts:

    r1 = sentiment1(text)[0]
    r2 = sentiment2(text)[0]

    print(f"Text: {text}")
    print(f"Model 1: {r1['label']} ({r1['score']:.3f})")
    print(f"Model 2: {r2['label']} ({r2['score']:.3f})")
    print()
```
