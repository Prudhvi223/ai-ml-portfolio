```python
from transformers import AutoTokenizer
```

```python
tokenizer = AutoTokenizer.from_pretrained(
    "distilbert-base-uncased"
)
```

```python
text = "Large language models are powerful."

tokens = tokenizer.tokenize(text)

print(tokens)
```

```python
token_ids = tokenizer.encode(text)

print(token_ids)
```

```python
from transformers import pipeline
```

```python
generator = pipeline(
    "text-generation",
    model="distilgpt2"
)
```

```python
prompt = "Artificial intelligence is"

result = generator(
    prompt,
    max_new_tokens=30
)

print(result[0]["generated_text"])
```

```python
prompts = [
    "Machine learning is",
    "The future of technology is",
    "Python is useful because"
]

for prompt in prompts:
    result = generator(
        prompt,
        max_new_tokens=25
    )

    print("Prompt:", prompt)
    print("Output:", result[0]["generated_text"])
    print("-" * 60)
```

```python
text = """
Large language models can understand and generate
human-like text using neural networks.
"""

tokens = tokenizer.tokenize(text)

print("Number of tokens:", len(tokens))
print("Tokens:", tokens)
```

```python
short_text = "AI is powerful."

long_text = """
Artificial intelligence is a field of computer science
that focuses on creating systems capable of performing
tasks that normally require human intelligence, such as
understanding language, recognizing patterns, reasoning,
and generating content.
"""

short_tokens = tokenizer.tokenize(short_text)
long_tokens = tokenizer.tokenize(long_text)

print("Short text tokens:", len(short_tokens))
print("Long text tokens:", len(long_tokens))
```

```python
question = "What is machine learning?"

context = """
Machine learning is a branch of artificial intelligence
where computers learn patterns from data and use those
patterns to make predictions or decisions.
"""

prompt = f"""
Context:
{context}

Question:
{question}

Answer:
"""

result = generator(
    prompt,
    max_new_tokens=50
)

print(result[0]["generated_text"])
```
