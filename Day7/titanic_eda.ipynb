# Titanic Exploratory Data Analysis (EDA)

This notebook explores the Titanic dataset using Pandas and Matplotlib.

Objectives:

- Understand the dataset
- Analyze missing values
- Study passenger survival
- Visualize important patterns

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../Day3/titanic.csv")

df.head()
```

## Dataset Preview

The first five rows give us an overview of the dataset. Each row represents one passenger on the Titanic.

```python
df.info()
```

## Dataset Information

The dataset contains 891 passengers and 12 columns. Some columns such as Age, Cabin and Embarked contain missing values.

```python
df.isnull().sum()
```

## Missing Values

Age contains some missing values.

Cabin has a large number of missing values.

Embarked has only a few missing values.

```python
df.describe()
```

## Statistical Summary

The average passenger age is approximately 29 years.

The average fare is around 32.

The summary also shows minimum, maximum and quartile values.

```python
df["Survived"].value_counts().plot(kind="bar")

plt.title("Survival Count")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Passengers")

plt.show()
```

## Observation

More passengers died than survived during the Titanic disaster.

## Survival Rate by Gender

This analysis compares the survival rate of male and female passengers.

```python
df.groupby("Sex")["Survived"].mean()
```

```python
df.groupby("Sex")["Survived"].mean().plot(kind="bar")

plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate")

plt.show()
```

### Observation

Female passengers had a significantly higher survival rate than male passengers.

## Survival Rate by Passenger Class

This analysis shows survival rates for different passenger classes.

```python
df.groupby("Pclass")["Survived"].mean()
```

```python
df.groupby("Pclass")["Survived"].mean().plot(kind="bar")

plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")

plt.show()
```

### Observation

Passengers traveling in First Class had the highest survival rate, while Third Class passengers had the lowest.

## Age Distribution

```python
plt.hist(df["Age"].dropna(), bins=20)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.show()
```

### Observation

Most passengers were between 20 and 40 years of age.

## Fare Distribution

```python
plt.hist(df["Fare"], bins=20)

plt.title("Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Number of Passengers")

plt.show()
```

### Observation

Most passengers paid relatively low fares, while only a small number paid very high fares.

# Conclusion

This exploratory data analysis provided several insights into the Titanic dataset.

## Key Findings

- The dataset contains 891 passenger records.
- Age and Cabin contain missing values.
- More passengers died than survived.
- Female passengers had a much higher survival rate than male passengers.
- First Class passengers survived more often than Third Class passengers.
- Most passengers were between 20 and 40 years old.
- The majority of passengers paid relatively low ticket fares.

This project demonstrates the use of Pandas and Matplotlib for exploratory data analysis.
