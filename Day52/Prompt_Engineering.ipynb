```python
from transformers import pipeline
```

```python
generator = pipeline(
    "text-generation",
    model="distilgpt2"
)
```

```python
prompt = """
Classify the following message as Spam or Ham.

Message:
Congratulations! You have won a free prize.

Answer:
"""

result = generator(
    prompt,
    max_new_tokens=20,
    do_sample=False
)

print(result[0]["generated_text"])
```

```python
prompt = """
Classify SMS messages as Spam or Ham.

Example 1:
Message: "Hey, are you coming to class?"
Answer: Ham

Example 2:
Message: "Congratulations! You won $1000!"
Answer: Spam

Example 3:
Message: "Can you send me the notes?"
Answer: Ham

Now classify:

Message: "You have won a free vacation!"
Answer:
"""

result = generator(
    prompt,
    max_new_tokens=20,
    do_sample=False
)

print(result[0]["generated_text"])
```

```python
prompt_patterns = {
    "zero_shot": """
Classify the following text as positive or negative:

Text: {text}

Answer:
""",

    "few_shot": """
Classify the text as positive or negative.

Example 1:
Text: I love this product.
Answer: Positive

Example 2:
Text: This product is terrible.
Answer: Negative

Now classify:
Text: {text}
Answer:
""",

    "summarization": """
Summarize the following text in 3 bullet points:

{text}
""",

    "explanation": """
Explain the following concept to a beginner using a simple example:

Concept: {concept}
""",

    "structured": """
Analyze the following text.

Return:
Topic:
Summary:
Key Point:

Text:
{text}
"""
}
```

```python
text = "Machine learning allows computers to learn patterns from data."

prompt = prompt_patterns["summarization"].format(
    text=text
)

print(prompt)
```

```python
text = "The new phone has excellent battery life and a great camera."
```

```python
zero_shot_prompt = f"""
Classify the sentiment of this text as Positive or Negative.

Text:
{text}

Answer:
"""

print(zero_shot_prompt)
```

```python
few_shot_prompt = f"""
Classify the sentiment as Positive or Negative.

Example 1:
Text: I love this product.
Answer: Positive

Example 2:
Text: This product is terrible.
Answer: Negative

Example 3:
Text: The service was excellent.
Answer: Positive

Now classify:

Text:
{text}

Answer:
"""

print(few_shot_prompt)
```

```python
structured_prompt = f"""
Analyze the sentiment of the following text.

Text:
{text}

Return exactly this format:

Sentiment:
Reason:
"""
    
print(structured_prompt)
```

```python
prompts = {
    "Zero-Shot": zero_shot_prompt,
    "Few-Shot": few_shot_prompt,
    "Structured": structured_prompt
}

for name, prompt in prompts.items():

    result = generator(
        prompt,
        max_new_tokens=40,
        do_sample=False,
        pad_token_id=generator.tokenizer.eos_token_id
    )

    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)
    print(result[0]["generated_text"])
```
