```python
import torch
import torch.nn.functional as F
```

```python
x = torch.tensor([2.0, 3.0])

w = torch.tensor([0.5, 0.4])

b = 1.0

z = torch.dot(x, w) + b

print(z)
```

```python
print(F.relu(z))
```

```python
print(torch.sigmoid(z))
```

```python
scores = torch.tensor([2.0,5.0,1.0])

print(F.softmax(scores, dim=0))
```
