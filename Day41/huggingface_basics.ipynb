```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained(
    "distilbert-base-uncased"
)
```

```python
text = "I love learning AI"

tokens = tokenizer.tokenize(text)

print(tokens)
```

```python
token_ids = tokenizer.convert_tokens_to_ids(tokens)

print(token_ids)
```

```python
from transformers import pipeline

sentiment = pipeline("sentiment-analysis")

result = sentiment(
    "I really enjoy learning deep learning."
)

print(result)
```

```python
texts = [
    "I love this course.",
    "This project is amazing.",
    "I hate this error.",
    "The movie was terrible."
]

results = sentiment(texts)

for text, result in zip(texts, results):
    print(text)
    print(result)
    print()
```

```python
from transformers import pipeline

qa = pipeline(
    "question-answering",
    model="distilbert-base-cased-distilled-squad"
)

context = """
Python is a popular programming language.
It is widely used in machine learning,
data science, and web development.
"""

question = "What is Python used for?"

answer = qa(
    question=question,
    context=context
)

print(answer)
```

```python
from transformers import pipeline

summarizer = pipeline("summarization")

text = """
Machine learning is a branch of artificial intelligence.
It allows computers to learn patterns from data.
These patterns can then be used to make predictions
or decisions without explicitly programming every rule.
"""

result = summarizer(
    text,
    max_length=40,
    min_length=10
)

print(result)
```

```python
from transformers import pipeline

sentiment = pipeline("sentiment-analysis")

texts = [
    "I love machine learning.",
    "This project is amazing.",
    "I don't like this error."
]

for text in texts:
    print(text)
    print(sentiment(text))
```

Tokenizer:
Converts text into tokens/numbers.

Pretrained model:
A model that has already been trained on large amounts of data.

Pipeline:
A simple Hugging Face interface that connects preprocessing,
model inference, and postprocessing.

Transfer learning:
Using knowledge learned by a pretrained model for another task.
