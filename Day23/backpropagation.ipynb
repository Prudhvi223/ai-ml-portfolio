```python
import torch
```

```python
x = torch.tensor(2.0, requires_grad=True)

y = x**2

print(y)
```

```python
y.backward()

print(x.grad)
```

```python
prediction = torch.tensor([2.5])

target = torch.tensor([3.0])

loss = torch.nn.functional.mse_loss(prediction, target)

print(loss)
```

```python
prediction = torch.tensor([[2.0,0.5]])

target = torch.tensor([0])

loss = torch.nn.CrossEntropyLoss()

print(loss(prediction,target))
```

```python
w = torch.tensor(5.0)

gradient = torch.tensor(2.0)

learning_rate = 0.1

new_w = w - learning_rate * gradient

print(new_w)
```

Loss measures prediction error.

Gradient is the slope.

Gradient Descent updates weights.

Backpropagation computes gradients.

PyTorch Autograd calculates gradients automatically.

MSE is used for regression.

CrossEntropy is used for classification.
