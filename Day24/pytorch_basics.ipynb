```python
import torch
import torch.nn as nn

print(torch.__version__)
```

```python
# Scalar
a = torch.tensor(5)

# Vector
b = torch.tensor([1, 2, 3])

# Matrix
c = torch.tensor([
    [1, 2],
    [3, 4]
])

print("Scalar:")
print(a)

print("\nVector:")
print(b)

print("\nMatrix:")
print(c)
```

```python
print("Shape of a:", a.shape)
print("Shape of b:", b.shape)
print("Shape of c:", c.shape)
```

```python
zeros = torch.zeros(2, 3)
ones = torch.ones(2, 3)
random_tensor = torch.rand(2, 3)

print("Zeros:")
print(zeros)

print("\nOnes:")
print(ones)

print("\nRandom:")
print(random_tensor)
```

```python
x = torch.tensor([1.0, 2.0, 3.0])
y = torch.tensor([4.0, 5.0, 6.0])

print("Addition:", x + y)
print("Subtraction:", x - y)
print("Multiplication:", x * y)
print("Division:", x / y)
```

```python
X = torch.tensor([
    [1.0, 2.0],
    [3.0, 4.0]
])

W = torch.tensor([
    [0.5, 0.2],
    [0.1, 0.7]
])

result = torch.matmul(X, W)

print(result)

ans= X@W

print(ans)
```

```python
x = torch.tensor(3.0, requires_grad=True)

y = x ** 2

y.backward()

print("x:", x)
print("y:", y)
print("dy/dx:", x.grad)
```

```python
x = torch.tensor(2.0, requires_grad=True)

y = 3 * x**2 + 2 * x + 1

y.backward()

print("y =", y.item())
print("Gradient =", x.grad.item())
```

```python
import torch
import torch.nn as nn
class SimpleNN(nn.Module):
    
    def __init__(self):
        super().__init__()
        
        self.layer1 = nn.Linear(2, 4)
        self.relu = nn.ReLU()
        self.layer2 = nn.Linear(4, 1)
    
    def forward(self, x):
        x = self.layer1(x)
        x = self.relu(x)
        x = self.layer2(x)
        
        return x
```

```python
model = SimpleNN()
print(model)
```

```python
input_data = torch.tensor([
    [1.0, 2.0]
])

output = model(input_data)

print("Input:")
print(input_data)

print("\nOutput:")
print(output)
```

```python
input_data = torch.tensor([
    [1.0, 2.0]
])

output = model(input_data)

print("Input:")
print(input_data)

print("\nOutput:")
print(output)
```

```python
for name, param in model.named_parameters():
    print(name)
    print(param)
    print()
```

```python
model = SimpleNN()

X = torch.tensor([
    [1.0, 2.0],
    [3.0, 4.0]
])

y_true = torch.tensor([
    [1.0],
    [0.0]
])

prediction = model(X)

loss = torch.mean((prediction - y_true) ** 2)

print("Prediction:")
print(prediction)

print("\nLoss:")
print(loss.item())
```
