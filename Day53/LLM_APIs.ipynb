```python
import ollama
from pydantic import BaseModel

class MessageInfo(BaseModel):
    category: str
    time: str
    day: str

response = ollama.chat(
    model="llama3.2:1b",
    messages=[
        {
            "role": "user",
            "content": "Extract information from this message: 'We have a meeting at 5 PM today.'"
        }
    ],
    format=MessageInfo.model_json_schema()
)

print(response["message"]["content"])
```

```python
result = MessageInfo.model_validate_json(
    response["message"]["content"]
)

print("Category:", result.category)
print("Time:", result.time)
print("Day:", result.day)
```

```python
def calculate_square(number: int) -> int:
    return number * number

print(calculate_square(8))
```

```python
import ollama

tools = [
    {
        "type": "function",
        "function": {
            "name": "calculate_square",
            "description": "Calculate the square of an integer.",
            "parameters": {
                "type": "object",
                "properties": {
                    "number": {
                        "type": "integer",
                        "description": "The integer to square."
                    }
                },
                "required": ["number"]
            }
        }
    }
]

response = ollama.chat(
    model="llama3.2:1b",
    messages=[
        {
            "role": "user",
            "content": "Calculate the square of 8 using the available tool."
        }
    ],
    tools=tools
)

print(response["message"])
```

```python
message = response["message"]

if message.get("tool_calls"):
    for tool_call in message["tool_calls"]:
        if tool_call["function"]["name"] == "calculate_square":
            number = tool_call["function"]["arguments"]["number"]

            result = calculate_square(number)

            print("Tool result:", result)

            final_response = ollama.chat(
                model="llama3.2:1b",
                messages=[
                    {
                        "role": "user",
                        "content": "Calculate the square of 8 using the available tool."
                    },
                    message,
                    {
                        "role": "tool",
                        "content": str(result)
                    }
                ],
                tools=tools
            )

            print(final_response["message"]["content"])
else:
    print("The model did not request a tool. Try running the previous cell again.")
```

```python
import ollama

def ask_llm(question):
    response = ollama.chat(
        model="llama3.2:1b",
        messages=[
            {
                "role": "system",
                "content": "You are a helpful AI assistant. Give clear, simple answers."
            },
            {
                "role": "user",
                "content": question
            }
        ]
    )

    return response["message"]["content"]

print(ask_llm("Explain the difference between AI and ML."))
```

```python
questions = [
    "What is an API?",
    "What is structured output?",
    "What is tool calling?"
]

for question in questions:
    print("\nQuestion:", question)
    print("Answer:", ask_llm(question))
```
